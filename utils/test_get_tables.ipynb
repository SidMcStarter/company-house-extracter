```python
def clean_unnamed_columns(df):
    """
    Clean DataFrame columns by replacing 'Unnamed' columns with empty strings.
    Works with both single-level and multi-level column indices.
    
    Args:
        df (pd.DataFrame): DataFrame with columns to clean
        
    Returns:
        pd.DataFrame: DataFrame with cleaned column names
    """
    # Create a copy to avoid modifying the original
    df_clean = df.copy()
    
    # Check if columns are multi-index
    if isinstance(df_clean.columns, pd.MultiIndex):
        # For multi-level columns
        new_columns = []
        for col in df_clean.columns:
            # Replace any 'Unnamed' components with empty string
            new_col = tuple('' if 'Unnamed' in str(c) else c for c in col)
            new_columns.append(new_col)
        
        df_clean.columns = pd.MultiIndex.from_tuples(new_columns)
    else:
        # For single-level columns
        df_clean.columns = ['' if 'Unnamed' in str(col) else col for col in df_clean.columns]
    
    return df_clean
```

```python
for table_id in tables:
    tables[table_id] = clean_unnamed_columns(tables[table_id])
```
